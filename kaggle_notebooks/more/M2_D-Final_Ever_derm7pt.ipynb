```
# ---- retrain + test-eval each tier-winning backbone at 512px, each in its own subprocess ----
import subprocess
import sys
import json as _json

RESULTS_DIR = f'{OUTPUT_DIR}/results'
os.makedirs(RESULTS_DIR, exist_ok=True)

results = []
overall_best = {'test_diagnosis_5way_acc': -1.0, 'backbone': None, 'tier': None}

for tier_name, backbones in BACKBONE_TIERS.items():
    print(f'\n========== TIER: {tier_name} ==========')
    for backbone_name in backbones:
        print(f'\n--- {backbone_name} (isolated subprocess, batch_size={BATCH_SIZES[tier_name]}) ---')
        result_path = f'{RESULTS_DIR}/{tier_name}_{backbone_name}.json'
        cmd = [sys.executable, f'{OUTPUT_DIR}/train_single_backbone.py',
               '--input-dir', INPUT_DIR, '--output-dir', OUTPUT_DIR,
               '--backbone', backbone_name, '--tier', tier_name,
               '--epochs', str(EPOCHS_PER_BACKBONE), '--patience', str(EARLY_STOP_PATIENCE),
               '--batch-size', str(BATCH_SIZES[tier_name]), '--lr', str(LR), '--img-size', str(IMG_SIZE),
               '--num-workers', str(NUM_WORKERS), '--result-path', result_path]
        for _attempt in range(2):
            proc = subprocess.run(cmd)
            if proc.returncode == 0 and os.path.exists(result_path):
                break
            note = ' -- retrying once in a fresh process' if _attempt == 0 else ' -- giving up after retry'
            print(f'  !! subprocess exited with code {proc.returncode} for {backbone_name} (see traceback above){note}')

        if os.path.exists(result_path):
            with open(result_path) as f:
                row = _json.load(f)
        else:
            row = {'tier': tier_name, 'backbone': backbone_name, 'error': 'no result file written'}
        results.append(row)

        acc5 = row.get('test_diagnosis_5way_acc')
        if acc5 is not None and acc5 > overall_best['test_diagnosis_5way_acc']:
            overall_best = {'test_diagnosis_5way_acc': acc5, 'backbone': backbone_name, 'tier': tier_name}
```
